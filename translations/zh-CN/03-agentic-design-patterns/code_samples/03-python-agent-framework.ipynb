在本课中，我们将探索构建高效 AI 代理的三个基础设计模式：

我们将把每种模式应用于一个<strong>旅游目的地推荐系统</strong>场景，逐步构建一个能够推荐目的地、检查可用性和处理物流的系统。

```python
import os
import dotenv
import asyncio
import logging
from pydantic import BaseModel
from agent_framework.openai import OpenAIChatClient
from agent_framework import tool
from typing import Annotated


dotenv.load_dotenv(dotenv.find_dotenv())
logging.getLogger("agent_framework.openai").setLevel(logging.ERROR)

api_key=os.getenv("OPENAI_API_KEY")
base_url=os.getenv("OPENAI_BASE_URL")
model=os.getenv("OPENAI_MODEL")

if not api_key or not base_url:
    raise ValueError(
        "Missing required environment variables."
        "Please set OPENAI_API_KEY and OPENAI_MODEL in your .env file."
    )
client=OpenAIChatClient(
    model=model,
    api_key=api_key,
    base_url=base_url
)
```

## 模式1：明确的代理指令

最有影响力的模式也是最简单的：为你的代理编写清晰、详细的指令。

- <strong>代理是谁</strong>（角色和语气）
- <strong>代理该做什么</strong>（逐步职责）
- <strong>代理应如何表现</strong>（约束和风格）

下面，我们创建一个旅行礼宾代理，带有明确的指令来塑造它生成的每个回复。

## 模式 2：使用 Pydantic 模型的结构化输出

自由形式文本对对话很有用，但下游系统需要结构化数据。
通过将 **Pydantic 模型** 与 <strong>工具函数</strong> 配对，我们可以：

- 定义代理输出的精确定义模式

- 可靠地将代理结果集成到应用逻辑中

模型返回一个经过验证的 `TravelRecommendations` 对象（可通过 `response.value` 访问）
，而不是自由格式文本。`get_destination_details` 工具也返回类型化的
`DestinationRecommendation`，因此数据从始至终保持结构化。

```python
class DestinationRecommendation(BaseModel):
    destination: str
    available: bool
    best_season: str
    highlights: list[str]
    estimated_budget_usd: int


class TravelRecommendations(BaseModel):
    recommendations: list[DestinationRecommendation]
    personalized_note: str


@tool(approval_mode="never_require")
def get_destination_details(
    destination: Annotated[str, "要查找的目的地"]
) -> DestinationRecommendation:
    """获取有关度假目的地的结构化详细信息。"""
    details = {
        "巴塞罗那": DestinationRecommendation(
            destination="Barcelona",
            available=True,
            best_season="5月-6月",
            highlights=["海滩", "建筑", "夜生活"],
            estimated_budget_usd=2000,
        ),
        "东京": DestinationRecommendation(
            destination="东京",
            available=True,
            best_season="3月-4月",
            highlights=["文化", "美食", "科技"],
            estimated_budget_usd=2500,
        ),
        "开普敦": DestinationRecommendation(
            destination="开普敦",
            available=False,
            best_season="11月-3月",
            highlights=["自然", "葡萄酒", "冒险"],
            estimated_budget_usd=1800,
        ),
    }
    return details.get(
        destination,
        DestinationRecommendation(
            destination=destination,
            available=False,
            best_season="Unknown",
            highlights=[],
            estimated_budget_usd=0,
        ),
    )


structured_agent = client.as_agent(
    name="结构化旅行专家",
    instructions="你是旅行专家。根据旅行者的喜好推荐目的地。使用 get_destination_details 工具。",
    tools=[get_destination_details],
)

# Passing `response_format` forces the agent to return a validated
# TravelRecommendations object instead of free-form text.
response = await structured_agent.run(
    "为预算为 2500 美元的文化爱好者推荐 3 个目的地",
    options={"response_format": TravelRecommendations},
)

if response and response.value:
    result: TravelRecommendations = response.value
    for rec in result.recommendations:
        status = "可用的" if rec.available else "无法使用"
        print(f"{rec.destination} ({status})")
        print(f"  最佳季节: {rec.best_season}")
        print(f"  亮点: {', '.join(rec.highlights)}")
        print(f"  预算: ${rec.estimated_budget_usd}")
        print()
    print(f"Note: {result.personalized_note}")
else:
    print("未返回经过验证的结构化响应。")
    print(response)
```

## 模式 3：单一职责代理

复杂任务通过将工作拆分为多个专注的代理来执行，每个代理负责单一职责：

- 一个了解地点和可用性的 <strong>目的地专家</strong>
- 一个处理航班、酒店和行程的 <strong>物流规划师</strong>

```python
destination_agent = client.as_agent(
    name="目的地专家",
    tools=[get_destination_details],
    instructions="""您是目的地研究专家。你唯一的工作就是：
1. 根据旅行者偏好评估目的地
2. 使用提供的工具检查可用性
3. 返回一个包含优点/缺点的简短排名列表
不要讨论航班、酒店或物流——其他代理会处理这些问题。""",
)

logistics_agent = client.as_agent(
    name="旅行行程规划师",
    instructions="""您是一名旅行行程规划师。你唯一的工作就是：
1. 为所选目的地创建每日行程
2. 在规定预算内建议航班和酒店选择
3. 注意签证要求和旅行保险建议
不要推荐目的地——另一个代理会处理这个问题。""",
)

# Step 1: Destination Expert picks the best options
dest_response = await destination_agent.run(
    "我想要 2500 美元以下的一周海滩和建筑体验。我应该去哪里？"
)
print("=== Destination Expert ===")
print(dest_response)

# Step 2: Logistics Planner builds the trip plan
logistics_response = await logistics_agent.run(
    f"根据此建议计划为期一周的旅行：\n{dest_response}"
)
print("\n=== Logistics Planner ===")
print(logistics_response)
```

## 总结

在本课中，我们将三个主动设计模式应用于旅游推荐场景：

| 模式 | 关键思想 | 优势 |
|---|---|---|
| <strong>明确指令</strong> | 预先定义角色、职责和约束 | 保持一致、符合品牌形象的代理行为 |
| <strong>结构化输出</strong> | 使用Pydantic模型作为响应格式 | 经过验证、机器可读的结果 |
| <strong>单一职责</strong> | 让每个代理专注于一项工作 | 更易测试、维护和组合 |

这些模式自然组合——你可以将明确指令与结构化输出结合到单一职责代理中，构建健壮、适合生产的系统。
